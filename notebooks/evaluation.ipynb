# Evaluate a trained grasp classifier

Replaces `tests.ipynb` of the thesis code. Scores a saved checkpoint on a result table (the `results.csv` format written by `smartrehab depth` / `smartrehab combine`) and writes F1 scores, the per-sample table and confusion matrices.

Same as: `smartrehab evaluate --checkpoint <model.pth> --data <table.csv>`.

```python
from pathlib import Path

from IPython.display import Image, display

from smartrehab.evaluate import evaluate_checkpoint

CHECKPOINT = Path("outputs/training/models/<model>.pth")  # produced by `smartrehab train`
DATA = Path("outputs/test_results.csv")                   # same columns as the training table
OUT = Path("outputs/evaluation")
```

```python
scores = evaluate_checkpoint(CHECKPOINT, DATA, OUT)
scores
```

```python
display(Image(OUT / "confusion_matrix_percent.png"))
display(Image(OUT / "confusion_matrix_counts.png"))
```
